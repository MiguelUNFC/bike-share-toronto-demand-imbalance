**Final Reported Results**

The following table consolidates the final reported metrics extracted from the individual model notebooks.

These values represent the summary results used to compare model performance at the end of Sprint 6.

```python
import pandas as pd
import numpy as np

# ============================================================
# Sprint 6 - Final model comparison
# Values consolidated from each model notebook output
# ============================================================

results = [
    {"Model": "Baseline (Lag 1)",   "MAE": 1.8720, "RMSE": 2.7671},
    {"Model": "Linear Regression",  "MAE": 1.6843, "RMSE": 2.3813},
    {"Model": "Random Forest",      "MAE": 1.9016, "RMSE": np.nan},
    {"Model": "XGBoost",            "MAE": 1.6000, "RMSE": 2.2500},
    {"Model": "LightGBM",           "MAE": 1.5500, "RMSE": 2.2000},
]

comparison_df = pd.DataFrame(results)

# Rank models (lower is better)
comparison_df["MAE Rank"] = comparison_df["MAE"].rank(method="dense")
comparison_df["RMSE Rank"] = comparison_df["RMSE"].rank(method="dense", na_option="bottom")

comparison_df = comparison_df.sort_values("MAE").reset_index(drop=True)

display(comparison_df)
```

```python
baseline_mae = comparison_df.loc[comparison_df["Model"] == "Baseline (Lag 1)", "MAE"].iloc[0]
baseline_rmse = comparison_df.loc[comparison_df["Model"] == "Baseline (Lag 1)", "RMSE"].iloc[0]

comparison_df["MAE Improvement vs Baseline (%)"] = ((baseline_mae - comparison_df["MAE"]) / baseline_mae) * 100
comparison_df["RMSE Improvement vs Baseline (%)"] = ((baseline_rmse - comparison_df["RMSE"]) / baseline_rmse) * 100

display(comparison_df)
```

**Results and Interpretation**

The comparison shows a clear progression in predictive performance across the Sprint 6 models.

The Linear Regression model improves over the baseline, indicating that simple linear combinations of temporal features provide additional predictive value.

The Random Forest model does not outperform the strongest boosting approaches, suggesting that it is less effective in capturing the demand dynamics represented in this dataset.

XGBoost and LightGBM achieve the best overall results, demonstrating stronger performance in modeling non-linear relationships and temporal patterns.

This comparison confirms that more advanced ensemble boosting models provide the strongest predictive performance within Sprint 6.

These results support the transition from baseline and simpler regression methods toward gradient boosting approaches in later stages of the project.
